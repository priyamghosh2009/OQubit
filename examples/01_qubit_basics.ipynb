# OQubit:- Qubits and StateVectors

This notebook introduces the two fundamental objects in OQubit:

- `Qubit` — represents a single quantum bit.
- `StateVector` — represents the combined state of one or more qubits.

No quantum gates are used in this example. We focus only on creating, inspecting, and combining quantum states.

## 1. Import OQubit

```python
import oqubit
from oqubit import Qubit, StateVector
```

````python
import oqubit
from oqubit import Qubit, StateVector
````

## 2. The computational basis states

A qubit can be in one of two computational basis states:

$|0⟩ =\begin{bmatrix} 
1 \\ 
0 
\end{bmatrix}
$

__and__

$
|1⟩ =
\begin{bmatrix}
0 \\
1 
\end{bmatrix}
$

OQubit provides convenient constructors for these states.
```python
q0 = Qubit.zero()
q1 = Qubit.one()
```
Inspect the states:
```python
print("Qubit 0:")
print(q0)

print("\nQubit 1:")
print(q1)
```

````python
q0 = Qubit.zero()
q1 = Qubit.one()
````

````python
print("Qubit 0:")
print(q0)

print("\nQubit 1:")
print(q1)
````

## 3. Qubit amplitudes

A general single-qubit state is written as

$
|\psi\rangle = \alpha|0\rangle + \beta|1\rangle
$

where α and β are complex probability amplitudes.

Let's create a qubit directly from its amplitudes:
```python
q = Qubit(alpha=(1 / 2**0.5),beta=(1 / 2**0.5))

print(q)
```
This represents

$
|\psi\rangle =
\frac{1}{\sqrt{2}}|0\rangle +
\frac{1}{\sqrt{2}}|1\rangle
$

````python
q = Qubit(alpha=(1 / 2**0.5),beta=(1 / 2**0.5))

print(q)
````

## 4. Inspecting the state

The state can be accessed as a NumPy array:

`q.state`

The two entries correspond to the amplitudes of |0⟩ and |1⟩:
```python
alpha = q.state[0]
beta = q.state[1]

print("α =", alpha)
print("β =", beta)
```

````python
alpha = q.state[0]
beta = q.state[1]

print("α =", alpha)
print("β =", beta)
````

## 5. Probabilities

The probability of measuring |0⟩ is related to the squared magnitude of α:

$
P(0) = |\alpha|^2
$

Similarly,
The probability of measuring |0⟩ is related to the squared magnitude of $\beta$:

$
P(1) = |\beta|^2
$

OQubit exposes these through prob_0 and prob_1.
```python
print("P(0) =", q.prob_0)
print("P(1) =", q.prob_1)
```
For the equal superposition above, both probabilities should be approximately:

__`0.5`__

Verify that the probabilities sum to one:

```python
print(q.prob_0 + q.prob_1)
```

````python
print("P(0) =", q.prob_0)
print("P(1) =", q.prob_1)
````

## 6. Creating different qubit states

OQubit can represent arbitrary normalized single-qubit states.

For example:
```python
q = Qubit(alpha=0.8,beta=0.6)
print(q)
print("P(0) =", q.prob_0)
print("P(1) =", q.prob_1)
```
Because

$
0.8$ <sup> 2 </sup>$ + 0.6 $<sup>2</sup>$ = 1
$

this state is already normalized.

````python
q = Qubit(alpha=0.8,beta=0.6)
print(q)
print("P(0) =", q.prob_0)
print("P(1) =", q.prob_1)
````

## 7. Complex amplitudes

Quantum amplitudes can also be complex numbers.
```python
q = Qubit(
    alpha=1 / 2**0.5,
    beta=1j / 2**0.5
)

print(q)
print("State:", q.state)
print("P(0):", q.prob_0)
print("P(1):", q.prob_1)
```
__Notice__ that the phase of an amplitude can be complex while the measurement probabilities still __depend on its magnitude__.

````python
q = Qubit(
    alpha=1 / 2**0.5,
    beta=1j / 2**0.5
)

print(q)
print("State:", q.state)
print("P(0):", q.prob_0)
print("P(1):", q.prob_1)
````

## 8. Creating a StateVector

A single qubit has a two-dimensional state vector.

For multiple qubits, the combined system is represented by a StateVector.

Create two qubits:
```python
q0 = Qubit.zero()
q1 = Qubit.plus()
```
Now combine them:
```python
state = StateVector([q0, q1])
```
Inspect the state:
```python
print(state)
```

````python
q0 = Qubit.zero()
q1 = Qubit.plus()
state = StateVector([q0, q1])
print(state)
````

## 9. Computational-basis notation

Two qubits have four computational basis states:

$
|00\rangle,\ |01\rangle,\ |10\rangle,\ |11\rangle
$

The corresponding state vector has four amplitudes:

$
|\psi\rangle =
\begin{bmatrix}
c_{00}\\
c_{01}\\
c_{10}\\
c_{11}
\end{bmatrix}
$

OQubit provides a notation representation:
```python
print(state.notation)
```
For example, two qubits initialized as |0⟩ and |1⟩ correspond to:

__|01⟩__

````python
q0 = Qubit.zero()
q1 = Qubit.one()
state = StateVector([q0, q1])
print(state)
print(state.notation)
````

## 10. Two-qubit basis states

Let's explicitly construct two basis-state qubits.
```python
q0 = Qubit.zero()
q1 = Qubit.plus()

state = StateVector([q0, q1])

print("State:")
print(state)

print("\nNotation:")
print(state.notation)
```
The combined system is formed using the tensor product:

$
|0\rangle \otimes |1\rangle = |01\rangle
$

````python
q0 = Qubit.zero()
q1 = Qubit.plus()

state = StateVector([q0, q1])

print("State:")
print(state)

print("\nNotation:")
print(state.notation)
````

## 11. Three qubits

The same idea works for larger systems.
```python
q1 = Qubit.zero()
q2 = Qubit.zero()
q3 = Qubit.plus()

state = StateVector([q1, q2, q3])

print(state)
print(state.notation)
```
Three qubits require:

$
2$<sup>3</sup>$ = 8
$

complex amplitudes.
```python
print("Number of amplitudes:", len(state.state))
```

````python
q1 = Qubit.zero()
q2 = Qubit.zero()
q3 = Qubit.plus()

state = StateVector([q1, q2, q3])

print(state)
print(state.notation)
````

## 12. Tensor-product growth

The dimension of a quantum state grows exponentially with the number of qubits.

| Qubits | Amplitudes |
|----------|------------|
| 1 | 2 |
| 2 | 4|
| 3 | 8 |
| 4	| 16 |
| 5	| 32 |
| 10 | 1,024 |
| 20 | 1,048,576 |

This is one of the fundamental computational challenges of classical quantum simulation.

For example:
```python
qubits = [Qubit.zero() for _ in range(5)]
state = StateVector(qubits)

print("Qubits:", len(qubits))
print("Amplitudes:", len(state.state))
```

````python
qubits = [Qubit.zero() for _ in range(5)]
state = StateVector(qubits)

print("Qubits:", len(qubits))
print("Amplitudes:", len(state.state))
````

## 13. Inspecting the StateVector

The underlying state can be accessed as an array:
```python
print(state.state)
```

For a five-qubit __|00010⟩__ state, the vector contains one __amplitude equal to 1__ and all __other amplitudes__ equal to __0__.

````python
q0=Qubit.zero()
q1=Qubit.one()
q=StateVector([q0,q0,q0,q1,q0])
print(q)
print(q.notation)
print("--------------------------------------")
print(q.state)
````
