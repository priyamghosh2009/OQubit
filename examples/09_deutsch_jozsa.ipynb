# Deutsch–Jozsa Algorithm

The Deutsch–Jozsa algorithm determines whether a Boolean function is **constant** or **balanced**.

For this example, the oracle is implemented with OQubit's `Circuit` API.

```python
from oqubit import Circuit
```

## Constant Oracle

We use the constant function $f(x)=0$.

The oracle applies no operation to the target qubit.

```python
n = 3
circuit = Circuit(n + 1)

# Prepare the auxiliary qubit in |1>
circuit.x(n)

# Create superposition
for qubit in range(n + 1):
    circuit.h(qubit)

# Constant oracle: f(x) = 0

# Apply interference to the input register
for qubit in range(n):
    circuit.h(qubit)

state = circuit.run()
result = circuit.measure(state=state, qubits=list(range(n)))

print("Measurement:", result)
print("Expected: all input qubits are 0")
```

## Balanced Oracle

Now use the balanced function

$$f(x)=x_0 \oplus x_1 \oplus \cdots \oplus x_{n-1}$$

A CNOT from every input qubit to the auxiliary qubit implements this oracle.

```python
n = 3
circuit = Circuit(n + 1)

# Prepare the auxiliary qubit in |1>
circuit.x(n)

# Create superposition
for qubit in range(n + 1):
    circuit.h(qubit)

# Balanced oracle
for qubit in range(n):
    circuit.cx(qubit, n)

# Apply interference to the input register
for qubit in range(n):
    circuit.h(qubit)

state = circuit.run()
result = circuit.measure(state=state, qubits=list(range(n)))

print("Measurement:", result)
print("Expected: a non-zero input state")
```

## Interpretation

After the final Hadamard gates, the input register gives:

- `000...0` → the function is **constant**
- any other state → the function is **balanced**

The Deutsch–Jozsa algorithm generalizes Deutsch's two-qubit algorithm to an arbitrary number of input qubits.

## Using the Deutsch-Jozsa Algorithm from OQubit

```python
from oqubit.algorithms import deutsch_jozsa
def sq(x):
    return x[1]
def cube(x):
    return x[0]
def aw_0(x):
    return 0
def aw_1(x):
    return 1
print(deutsch_jozsa(aw_1,1))
print(deutsch_jozsa(cube,2))
print(deutsch_jozsa(sq,2))
print(deutsch_jozsa(aw_0,1))
```
