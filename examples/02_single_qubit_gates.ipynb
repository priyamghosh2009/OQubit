## 2. Starting with the `|0⟩` state

We begin with the computational basis state:

$$
|0\rangle =
\begin{bmatrix}
1\\
0
\end{bmatrix}
$$

```python
q = Qubit.zero()

print("Initial state:")
print(q)

print("\nState vector:")
print(q.state)

print("\nP(0):", q.prob_0)
print("P(1):", q.prob_1)
```

````python
q = Qubit.zero()

print("Initial state:")
print(q)

print("\nState vector:")
print(q.state)

print("\nP(0):", q.prob_0)
print("P(1):", q.prob_1)
````

## 3. Pauli-X Gate

The Pauli-X gate is the quantum equivalent of a bit flip.

$$
X =
\begin{bmatrix}
0 & 1\\
1 & 0
\end{bmatrix}
$$

It transforms:

$$
X|0\rangle = |1\rangle
$$

and

$$
X|1\rangle = |0\rangle
$$

Apply `X` to `|0⟩`:

```python
q = Qubit.zero()

q.apply(oqubit.X)

print(q)
print(q.state)
```

Check the probabilities:

```python
print("P(0):", q.prob_0)
print("P(1):", q.prob_1)
```

The qubit should now be in `|1⟩`.

````python
q = Qubit.zero()

q.apply(oqubit.X)

print(q)
print(q.state)
````

## 4. Applying X to `|1⟩`

Create the `|1⟩` state using the appropriate OQubit constructor.

```python
q = Qubit.one()

q.apply(oqubit.X)

print(q)
print(q.state)
```

Applying X again returns the qubit to `|0⟩`.


````python
q = Qubit.one()

q.apply(oqubit.X)

print(q)
print(q.state)
````

## 5. Pauli-Y Gate

The Pauli-Y gate is:

$$
Y =
\begin{bmatrix}
0 & -i\\
i & 0
\end{bmatrix}
$$

Apply it to `|0⟩`:

```python
q = Qubit.zero()

q.apply(oqubit.Y)

print(q)
print(q.state)
```

The resulting state contains a complex amplitude.

Check its probabilities:

```python
print("P(0):", q.prob_0)
print("P(1):", q.prob_1)
```

Although the amplitudes contain a phase, the measurement probabilities remain normalized.


````python
q = Qubit.zero()

q.apply(oqubit.Y)

print(q)
print(q.state)
````

````python
q = Qubit.one()

q.apply(oqubit.Y)

print(q)
print(q.state)
````

## 6. Pauli-Z Gate

The Pauli-Z gate is:

$$
Z =
\begin{bmatrix}
1 & 0\\
0 & -1
\end{bmatrix}
$$

It leaves `|0⟩` unchanged:

$$
Z|0\rangle = |0\rangle
$$

and changes the phase of `|1⟩`:

$$
Z|1\rangle = -|1\rangle
$$

Try it on `|0⟩`:

```python
q = Qubit.zero()

q.apply(oqubit.Z)

print(q)
print(q.state)
```

Now try it on `|1⟩`:

```python
q = Qubit.one()
q.apply(oqubit.Z)
print(q)
print(q.state)
```

Notice the change in amplitude phase.


````python
q = Qubit.zero()

q.apply(oqubit.Z)

print(q)
print(q.state)
````

````python
q = Qubit.one()
q.apply(oqubit.Z)
print(q)
print(q.state)
````

## 7. Hadamard Gate

The Hadamard gate is one of the most important quantum gates.

$$
H =
\frac{1}{\sqrt{2}}
\begin{bmatrix}
1 & 1\\
1 & -1
\end{bmatrix}
$$

It transforms `|0⟩` into an equal superposition:

$$
H|0\rangle =
\frac{|0\rangle + |1\rangle}{\sqrt{2}}
$$

Apply H:

```python
q = Qubit.zero()

q.apply(oqubit.H)

print(q)
print(q.state)
```

Check the probabilities:

```python
print("P(0):", q.prob_0)
print("P(1):", q.prob_1)
```

Both probabilities should be approximately:

```text
0.5
```


````python
q = Qubit.zero()

q.apply(oqubit.H)

print(q)
print(q.state)
````

````python
q = Qubit.one()

q.apply(oqubit.H)

print(q)
print(q.state)
````

## 8. Hadamard applied twice

The Hadamard gate is its own inverse:

$$
H^2 = I
$$

Therefore:

$$
H(H|0\rangle) = |0\rangle
$$

Test this with OQubit:

```python
q = Qubit.zero()
q.apply(oqubit.H)
q.apply(oqubit.H)
print(q)
print(q.state)
```

The qubit should return to `|0⟩`.


````python
q = Qubit.zero()
q.apply(oqubit.H)
q.apply(oqubit.H)
print(q)
print(q.state)
````

## 9. Phase-S Gate

The S gate is a phase gate:

$$
S =
\begin{bmatrix}
1 & 0\\
0 & i
\end{bmatrix}
$$

Apply it to `|1⟩`:

```python
q = Qubit.one()

q.apply(oqubit.S)

print(q)
print(q.state)
```

The probability remains unchanged:

```python
print("P(0):", q.prob_0)
print("P(1):", q.prob_1)
```

The important change is the **phase** of the `|1⟩` amplitude.

````python
q = Qubit.one()

q.apply(oqubit.S)

print(q)
print(q.state)
````

## 10. T Gate

The T gate applies a phase of:

$$
\frac{\pi}{4}
$$

Its matrix is:

$$
T =
\begin{bmatrix}
1 & 0\\
0 & e^{i\pi/4}
\end{bmatrix}
$$

Apply it to `|1⟩`:

```python
q = Qubit.one()

q.apply(oqubit.T)

print(q)
print(q.state)
```

Check the probabilities:

```python
print("P(0):", q.prob_0)
print("P(1):", q.prob_1)
```

Again, the measurement probabilities remain unchanged because the T gate changes the phase rather than the amplitude magnitude.



````python
q = Qubit.one()

q.apply(oqubit.T)

print(q)
print(q.state)
````

## 12. A sequence of gates

Quantum programs are built by applying sequences of operations.

Start with `|0⟩`:

```python
q = Qubit.zero()
```

Apply H:

```python
q.apply(oqubit.H)
```

Then Z:

```python
q.apply(oqubit.Z)
```

Then H again:

```python
q.apply(oqubit.H)
```

Inspect the result:

```python
print(q)
print(q.state)

print("\nP(0):", q.prob_0)
print("P(1):", q.prob_1)
```

The sequence

$$
H \rightarrow Z \rightarrow H
$$

is equivalent to the Pauli-X operation:

$$
HZH = X
$$


````python
q = Qubit.zero()
q.apply(oqubit.H)
q.apply(oqubit.Z)
q.apply(oqubit.H)
print(q)
print(q.state)

print("\nP(0):", q.prob_0)
print("P(1):", q.prob_1)
````

## 13. Gate composition

We can also examine a simple sequence:

```python
q = Qubit.zero()
q.apply(oqubit.H)
q.apply(oqubit.X)
print(q)
print(q.state)
```

The state after H is:

$$
\frac{|0\rangle + |1\rangle}{\sqrt{2}}
$$

Applying X exchanges the amplitudes:

$$
\frac{|1\rangle + |0\rangle}{\sqrt{2}}
$$

which is physically the same equal-superposition state.

````python
q = Qubit.zero()
q.apply(oqubit.H)
q.apply(oqubit.X)
print(q)
print(q.state)
````

## 14. Returning to the original state

The Pauli-X gate is also self-inverse:

$$
X^2 = I
$$

```python
q = Qubit.zero()
q.apply(oqubit.X)
q.apply(oqubit.X)
print(q)
print(q.state)
```

The original `|0⟩` state is restored.

````python
q = Qubit.zero()
q.apply(oqubit.X)
q.apply(oqubit.X)
print(q)
print(q.state)
````

## 15. Summary

We explored the fundamental single-qubit gates in OQubit.

| Gate | Operation                  |
| ---- | -------------------------- |
| X    | Bit flip                   |
| Y    | Bit + phase transformation |
| Z    | Phase flip                 |
| H    | Creates superposition      |
| S    | π/2 phase                  |
| T    | π/4 phase                  |

The gates operate on the two-dimensional qubit state:

$$
|\psi\rangle =
\begin{bmatrix}
\alpha\\
\beta
\end{bmatrix}
$$

and transform it according to:

$$
|\psi'\rangle = U|\psi\rangle
$$

where `U` is the gate's unitary matrix.
