## 2. Computational Basis States

A qubit has two computational basis states:

$$
|0\rangle =
\begin{bmatrix}
1\\
0
\end{bmatrix}
$$

and

$$
|1\rangle =
\begin{bmatrix}
0\\
1
\end{bmatrix}
$$

Create the `|0⟩` state:

```python
q0 = Qubit.zero()

print(q0)
print(q0.state)
```

Create the `|1⟩` state using the constructor provided by your current OQubit version:

```python

print(q1.state)
```

$$
|\psi\rangle = \alpha|0\rangle + \beta|1\rangle

$$

```python
print("P(0) =", q0.prob_0)
print("P(1) =", q0.prob_1)
```

For `|1⟩`:

```python
print("P(0) =", q1.prob_0)
print("P(1) =", q1.prob_1)
```

````python
print("P(0) =", q1.prob_0)
print("P(1) =", q1.prob_1)
````

# 4. A General Quantum State

A general single-qubit state is:

$$
|\psi\rangle = \alpha|0\rangle+\beta|1\rangle
$$

with the normalization condition:

$$
|\alpha|^2+|\beta|^2=1
$$

Let's construct:

$$
|\psi\rangle =
\frac{3}{5}|0\rangle+
\frac{4}{5}|1\rangle
$$

```python
q = Qubit(
    alpha=3 / 5,
    beta=4 / 5
)

print(q)
print(q.state)
```

Check normalization:

```python
print(
    abs(q.state[0]) ** 2 +
    abs(q.state[1]) ** 2
)
```

The result should be:

```text
1.0
```


````python
q = Qubit(
    alpha=3 / 5,
    beta=4 / 5
)

print(q)
print(q.state)
````

````python
print(
    abs(q.state[0]) ** 2 +
    abs(q.state[1]) ** 2
)
````

## 5. Measuring the State Probabilities

```python
print("P(0) =", q.prob_0)
print("P(1) =", q.prob_1)
```

For this state:

$$
P(0)=\left|\frac35\right|^2=\frac9{25}
$$

and

$$
P(1)=\left|\frac45\right|^2=\frac{16}{25}
$$

Therefore:

```python
print("Total probability =", q.prob_0 + q.prob_1)
```

The total probability is `1`.

````python
print("Total probability =", q.prob_0 + q.prob_1)
````

## 6. Equal Superposition

One of the most important quantum states is the equal superposition:

$$
|+\rangle =

Create it directly:

```python
plus = Qubit.plus()
```

Check the probabilities:

```python
print("P(0) =", plus.prob_0)
```

Both should be approximately:

````python
plus = Qubit.plus()

print(plus)
print(plus.state)
````

````python
print("P(0) =", plus.prob_0)
print("P(1) =", plus.prob_1)
````

## 7. The Negative Superposition

Another important state is:

$$
|-\rangle =
\frac{|0\rangle-|1\rangle}{\sqrt2}
$$

Create it with OQubit:

```python
minus = Qubit.minus()

print(minus.state)
```

Check its probabilities:

```python
print("P(0) =", minus.prob_0)
```

Both probabilities are still `0.5`.

````python
print("P(0) =", minus.prob_0)
print("P(1) =", minus.prob_1)
````

## 8. Complex Quantum States

Quantum amplitudes can be complex numbers.

Consider:

$$
|\psi\rangle =
\frac{1}{\sqrt2}|0\rangle+
\frac{i}{\sqrt2}|1\rangle
$$

```python
q = Qubit(
    alpha=1 / 2**0.5,
    beta=1j / 2**0.5
)

print(q)
print(q.state)
```

The probabilities are:

```python
print("P(0) =", q.prob_0)
print("P(1) =", q.prob_1)
```

Even though the second amplitude is imaginary, its probability is determined by its magnitude:

$$
|i/\sqrt2|^2=\frac12
$$


````python
q = Qubit(
    alpha=1 / 2**0.5,
    beta=1j / 2**0.5
)

print(q)
print(q.state)
````

````python
print("P(0) =", q.prob_0)
print("P(1) =", q.prob_1)
````

## 9. Relative Phase

Compare these two states:

$$
|\psi_1\rangle =
\frac{|0\rangle+|1\rangle}{\sqrt2}
$$

and

$$
|\psi_2\rangle =
\frac{|0\rangle-|1\rangle}{\sqrt2}
$$

Create them:

```python
plus = Qubit.plus()
minus = Qubit.minus()

print("|+> :", plus.state)
print("|-> :", minus.state)
```

Their computational-basis probabilities are identical:

```python
print("Plus :", plus.prob_0, plus.prob_1)
print("Minus:", minus.prob_0, minus.prob_1)
```

But they are different quantum states because their amplitudes have different relative phases.

This distinction becomes important when quantum gates cause amplitudes to interfere.


````python
plus = Qubit.plus()
minus = Qubit.minus()

print("|+> :", plus.state)
print("|-> :", minus.state)
````

````python
print("Plus :", plus.prob_0, plus.prob_1)
print("Minus:", minus.prob_0, minus.prob_1)
````

## 10. Global Phase

A global phase multiplies every amplitude by the same complex number:

$$
|\psi'\rangle=e^{i\phi}|\psi\rangle
$$

For example:

$$
|\psi'\rangle=i|\psi\rangle
$$

The measurement probabilities remain unchanged.

```python
import numpy as np

q = Qubit.plus()

global_phase = 1j

phased_state = global_phase * q.state

print("Original:")
print(q.state)

print("\nGlobal-phase transformed:")
print(phased_state)
```

Compare the probabilities:

```python
print("Original probabilities:")
print(q.prob_0, q.prob_1)

print("\nAfter global phase:")
print(abs(phased_state[0]) ** 2,
      abs(phased_state[1]) ** 2)
```

The probabilities are unchanged.


````python
import numpy as np

q = Qubit.plus()

global_phase = 1j

phased_state = global_phase * q.state

print("Original:")
print(q.state)

print("\nGlobal-phase transformed:")
print(phased_state)
````

````python
print("Original probabilities:")
print(q.prob_0, q.prob_1)

print("\nAfter global phase:")
print(abs(phased_state[0]) ** 2,
      abs(phased_state[1]) ** 2)
````

## 11. Normalization

A valid quantum state must satisfy:

$$
\sum_i |c_i|^2=1
$$

OQubit can normalize states when constructing a `Qubit`.

For example, the amplitudes:

```python
alpha = 3
beta = 4
```

are not normalized:

$$
3^2+4^2=25
$$

```python
q = Qubit(
    alpha=3,
    beta=4,
    normalize=True
)

print(q.state)
```

Check:

```python
print(
    abs(q.state[0]) ** 2 +
    abs(q.state[1]) ** 2
)
```

The result should be `1`.


````python
q = Qubit(
    alpha=3,
    beta=4,
    normalize=True
)

print(q.state)
````

````python
print(
    abs(q.state[0]) ** 2 +
    abs(q.state[1]) ** 2
)
````

## 13. StateVector Representation

A single qubit has two amplitudes.

Two qubits require four amplitudes.

Three qubits require eight amplitudes.

Create a two-qubit computational basis state:

```python
state = StateVector([
    Qubit.zero(),
    Qubit.one()
])

print(state)
print(state.state)
print(state.notation)
```

This corresponds to:

$$
|01\rangle
$$

````python
state = StateVector([
    Qubit.zero(),
    Qubit.one()
])

print(state)
print(state.state)
print(state.notation)
````

## 14. A Three-Qubit State

Create:

```python
state = StateVector([
    Qubit.zero(),

])

```

$$
2^3=8


## 15. Superposition in a Multi-Qubit System

Create two `|+⟩` qubits:

```python
q1 = Qubit.plus()
q2 = Qubit.plus()

state = StateVector([q1, q2])

print(state.state)
print(state.notation)
```

Mathematically:

$$
|+\rangle\otimes|+\rangle
$$

gives:

$$
\frac{1}{2}
(
|00\rangle+
|01\rangle+
|10\rangle+
|11\rangle
)
$$

Each computational basis state therefore has probability:

$$
\frac14
$$

````python
q1 = Qubit.plus()
q2 = Qubit.plus()

state = StateVector([q1, q2])

print(state.state)
print(state.notation)
````

## 16. State Dimension

The number of amplitudes increases exponentially:

```python
for n in range(1, 11):
    print(f"{n} qubits -> {2**n:,} amplitudes")
```

Expected output:

```text
1 qubits -> 2 amplitudes
2 qubits -> 4 amplitudes
3 qubits -> 8 amplitudes
4 qubits -> 16 amplitudes
5 qubits -> 32 amplitudes
6 qubits -> 64 amplitudes
7 qubits -> 128 amplitudes
8 qubits -> 256 amplitudes
9 qubits -> 512 amplitudes
10 qubits -> 1,024 amplitudes
```

This exponential growth is one of the central challenges in classical quantum simulation.


````python
for n in range(1, 11):
    print(f"{n} qubits -> {2**n:,} amplitudes")
````

## 17. State Summary

The fundamental ideas covered in this notebook are:

### Computational basis

$$
|0\rangle,\quad |1\rangle
$$

### Superposition

$$
|\psi\rangle=\alpha|0\rangle+\beta|1\rangle
$$

### Normalization

$$
|\alpha|^2+|\beta|^2=1
$$

### Measurement probability

$$
P(0)=|\alpha|^2
$$

$$
P(1)=|\beta|^2
$$

### Multi-qubit states

$$
|\psi\rangle=
|\psi_1\rangle\otimes|\psi_2\rangle\otimes\cdots
$$

and an `n`-qubit state requires:

$$
2^n
$$

complex amplitudes.
