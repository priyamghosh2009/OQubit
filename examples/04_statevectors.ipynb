# OQubit — StateVectors

This notebook explores the `StateVector` class in OQubit.

A state vector is the mathematical representation of a quantum system. It stores the complex probability amplitudes associated with the computational-basis states of the system.

We will explore:

* Single-qubit state vectors
* Multi-qubit state vectors
* Tensor products
* Computational-basis ordering
* State-vector dimensions
* Amplitudes
* Probabilities
* Normalization
* Superposition
* Multi-qubit state construction

## 2. What is a StateVector?

A single qubit can be represented as:

$$
|\psi\rangle =
\alpha|0\rangle + \beta|1\rangle
$$

Its state vector is:

$$
|\psi\rangle =
\begin{bmatrix}
\alpha\\
\beta
\end{bmatrix}
$$

For example:

$$
|0\rangle =
\begin{bmatrix}
1\\
0
\end{bmatrix}
$$

Let's create a single-qubit state:

```python
q = Qubit.zero()

state = StateVector([q])

print(state)
print(state.state)
```

````python
q = Qubit.zero()

state = StateVector([q])

print(state)
print(state.state)
````

## 3. StateVector Dimension

A single qubit has:

$$
2^1 = 2
$$

complex amplitudes.

```python
print("Number of qubits:", 1)
print("Number of amplitudes:", len(state.state))
```

The state vector is therefore two-dimensional.


````python
print("Number of qubits:", 1)
print("Number of amplitudes:", len(state.state))
````

## 5. Computational-Basis Ordering

For two qubits, the computational basis is conventionally ordered as:

$$
|00\rangle,\ |01\rangle,\ |10\rangle,\ |11\rangle
$$

Therefore, a general two-qubit state can be written:

$$
|\psi\rangle =
c_{00}|00\rangle+
c_{01}|01\rangle+
c_{10}|10\rangle+
c_{11}|11\rangle
$$

OQubit exposes the state-vector representation:

```python
state = StateVector([
    Qubit.zero(),
    Qubit.one()
])

print("Notation:")
print(state.notation)

print("\nState vector:")
print(state.state)
```

The non-zero amplitude corresponds to:

$$
|01\rangle
$$


````python
state = StateVector([
    Qubit.zero(),
    Qubit.one()
])

print("Notation:")
print(state.notation)

print("\nState vector:")
print(state.state)
````

## 6. Tensor Products

Multi-qubit state vectors are constructed using tensor products.

For two qubits:

$$
|\psi\rangle =
|\psi_1\rangle\otimes|\psi_2\rangle
$$

Consider:

$$
|0\rangle =
\begin{bmatrix}
1\\
0
\end{bmatrix}
$$

and:

$$
|1\rangle =
\begin{bmatrix}
0\\
1
\end{bmatrix}
$$

Their tensor product is:

$$
|0\rangle\otimes|1\rangle =
\begin{bmatrix}
0\\
1\\
0\\
0
\end{bmatrix}
$$

OQubit performs this combination when constructing a `StateVector`.

```python
q0 = Qubit.zero()
q1 = Qubit.one()

state = StateVector([q0, q1])

print(state.state)
```


````python
q0 = Qubit.zero()
q1 = Qubit.one()

state = StateVector([q0, q1])

print(state.state)
````

## 10. Basis-State Probabilities

For a state vector:

$$
|\psi\rangle =
\sum_i c_i|i\rangle
$$

the probability associated with basis state `|i⟩` is:

$$
P(i)=|c_i|^2
$$

We can calculate the probabilities from the state vector:

```python
probabilities = np.abs(state.state) ** 2

print(probabilities)
```

Check that they sum to one:

```python
print("Total probability:", probabilities.sum())
```


````python
probabilities = np.abs(state.state) ** 2
print(probabilities)
print("Total probability:", probabilities.sum())
````

## 11. Equal Superposition of Two Qubits

Create two `|+⟩` states:

```python
q1 = Qubit.plus()
q2 = Qubit.plus()

state = StateVector([q1, q2])

print(state.state)
print(state.notation)
```

Mathematically:

$$
|+\rangle\otimes|+\rangle
$$

produces:

$$
\frac{1}{2}
(
|00\rangle+
|01\rangle+
|10\rangle+
|11\rangle
)
$$

Each basis state therefore has probability:

$$
\frac14
$$

Calculate the probabilities:

```python
probabilities = np.abs(state.state) ** 2

print(probabilities)
print("Total:", probabilities.sum())
```


````python
q1 = Qubit.plus()
q2 = Qubit.plus()

state = StateVector([q1, q2])

print(state.state)
print(state.notation)

probabilities = np.abs(state.state) ** 2

print(probabilities)
print("Total:", probabilities.sum())
````

## 15. Memory Requirements

If each amplitude is represented using a complex double-precision number, each amplitude requires approximately 16 bytes.

The approximate memory requirement is:

$$
M = 16\times2^n
$$

bytes.

```python
for n in range(10, 31, 5):
    amplitudes = 2 ** n
    memory = amplitudes * 16

    print(
        f"{n:2} qubits: "
        f"{amplitudes:,} amplitudes, "
        f"{memory / (1024**2):.2f} MiB"
    )
```

This illustrates why increasing the number of simulated qubits quickly becomes expensive.


````python
for n in range(10, 31, 5):
    amplitudes = 2 ** n
    memory = amplitudes * 16

    print(
        f"{n:2} qubits: "
        f"{amplitudes:,} amplitudes, "
        f"{memory / (1024**2):.2f} MiB"
    )
````

## 16. Complex Amplitudes

State vectors are not restricted to real numbers.

Consider:

$$
|\psi\rangle =
\frac{1}{\sqrt2}|0\rangle+
\frac{i}{\sqrt2}|1\rangle
$$

```python
q = Qubit(
    alpha=1 / np.sqrt(2),
    beta=1j / np.sqrt(2)
)

state = StateVector([q])

print(state.state)
```

The amplitudes are complex:

```python
for amplitude in state.state:
    print(amplitude)
```


````python
q = Qubit(
    alpha=1 / np.sqrt(2),
    beta=1j / np.sqrt(2)
)

state = StateVector([q])

print(state.state)

for amplitude in state.state:
    print(amplitude)
````

## 17. Normalization of a StateVector

A valid quantum state satisfies:

$$
\langle\psi|\psi\rangle=1
$$

For a state vector this means:

$$
\sum_i |c_i|^2=1
$$

Check the norm:

```python
norm = np.linalg.norm(state.state)

print("Norm:", norm)
```

A normalized state should have a norm of approximately:

```text
1.0
```


````python
norm = np.linalg.norm(state.state)

print("Norm:", norm)
````

## 18. Basis-State Indexing

The amplitudes correspond to computational-basis states.

For two qubits:

```python
basis_states = [
    "|00⟩",
    "|01⟩",
    "|10⟩",
    "|11⟩"
]

state = StateVector([
    Qubit.plus(),
```

This provides a direct connection between the numerical vector and its quantum-mechanical notation.


````python
basis_states = [
    "|00⟩",
    "|01⟩",
    "|10⟩",
    "|11⟩"
]

state = StateVector([
    Qubit.plus(),
    Qubit.plus()
])

for basis, amplitude in zip(basis_states, state.state):
    print(f"{basis}: {amplitude}")
````

## 20. StateVector Summary

The `StateVector` is the central numerical representation of a quantum system.

For `n` qubits:

$$
|\psi\rangle =
\sum_{i=0}^{2^n-1}c_i|i\rangle
$$

where:

* `cᵢ` is a complex probability amplitude.
* `|cᵢ|²` is the probability of the corresponding computational-basis state.
* The state must satisfy:

$$
\sum_i|c_i|^2=1
$$

The number of amplitudes is:

$$
2^n
$$
