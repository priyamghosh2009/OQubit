# Deutsch Algorithm

Deutsch's algorithm determines whether a one-bit Boolean function is constant or balanced using a single oracle evaluation.

This example uses OQubit's public `Circuit` API.

```python
from oqubit import Circuit
```

## Constant Oracle

For the constant function $f(x)=0$, the oracle does not modify the target qubit.

```python
circuit = Circuit(2)

# Prepare |0>|1>
circuit.x(1)

# Create superposition
circuit.h(0)
circuit.h(1)

# Constant oracle: f(x) = 0

# Interference
circuit.h(0)

state = circuit.run()
result = circuit.measure(state=state, qubits=0)

print("Measurement:", result)
print("Constant oracle")
```

## Balanced Oracle

For the balanced function $f(x)=x$, the oracle is a controlled-X operation.

```python
circuit = Circuit(2)

# Prepare |0>|1>
circuit.x(1)

# Create superposition
circuit.h(0)
circuit.h(1)

# Balanced oracle: f(x) = x
circuit.cx(0, 1)

# Interference
circuit.h(0)

state = circuit.run()
result = circuit.measure(state=state, qubits=0)

print("Measurement:", result)
print("Balanced oracle")
```

## Interpretation

The first qubit contains the result after the final Hadamard:

- `0` → the function is **constant**
- `1` → the function is **balanced**

The algorithm therefore distinguishes the two cases with one oracle evaluation.

## Using Deutsch algorithm directly from OQubit

```python
from oqubit.algorithms import deutsch
def sq(x):
    return x**2
def cube(x):
    return x**3
def aw_0(x):
    return 0
def aw_1(x):
    return 1
print(deutsch(aw_1))
print(deutsch(cube))
print(deutsch(sq))
print(deutsch(aw_0))
```
