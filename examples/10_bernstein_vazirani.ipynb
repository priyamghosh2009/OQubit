# Bernstein–Vazirani Algorithm

The Bernstein–Vazirani algorithm finds a hidden binary string
using a quantum oracle.

This example uses OQubit's built-in algorithm implementation.


```python
import oqubit
from oqubit.algorithms import bernstein_vazirani
```

## Run the Algorithm

Define a secret string and recover it using the algorithm.

```python
secret = "1011"

result = bernstein_vazirani(secret)

print("Secret:   ", secret)
print("Result:   ", result)
```

## Verify

The algorithm should recover the original secret string.

```python
assert result == secret

print("Bernstein–Vazirani test passed!")
```
