# Bell States

This notebook explores the four **Bell states**, the fundamental maximally entangled two-qubit states.

We will construct:

$
|\Phi^+\rangle =
\frac{|00\rangle + |11\rangle}{\sqrt{2}}
\
$

$
|\Phi^-\rangle =
\frac{|00\rangle - |11\rangle}{\sqrt{2}}
$

$
|\Psi^+\rangle =
\frac{|01\rangle + |10\rangle}{\sqrt{2}}
$

$
|\Psi^-\rangle =
\frac{|01\rangle - |10\rangle}{\sqrt{2}}
$
We will use OQubit to:

- Create computational-basis states
- Apply Hadamard and Pauli gates
- Apply controlled-X (CNOT)
- Construct all four Bell states
- Inspect their state vectors
- Verify their mathematical representations
- Understand entanglement

````python
import numpy as np
import oqubit

from oqubit import Qubit, StateVector
````

## 1. What Are Bell States?

Bell states are four special two-qubit quantum states.

They are maximally entangled, meaning that the measurement results of the two qubits are strongly correlated.

The four Bell states are:

$
|\Phi^+\rangle =
\frac{|00\rangle + |11\rangle}{\sqrt{2}}
$

$
|\Phi^-\rangle =
\frac{|00\rangle - |11\rangle}{\sqrt{2}}
$

$
|\Psi^+\rangle =
\frac{|01\rangle + |10\rangle}{\sqrt{2}}
$

$
|\Psi^-\rangle =
\frac{|01\rangle - |10\rangle}{\sqrt{2}}
$

## 2. Computational-Basis States

A two-qubit system has four computational-basis states:

$
|00\rangle,\quad |01\rangle,\quad |10\rangle,\quad |11\rangle
$

We will use these states as the starting point for constructing the Bell states.

````python
zero = Qubit.zero()
one = Qubit(alpha=0, beta=1)

state_00 = StateVector([zero, zero])
state_01 = StateVector([zero, one])
state_10 = StateVector([one, zero])
state_11 = StateVector([one, one])

print(state_00.notation)
print(state_01.notation)
print(state_10.notation)
print(state_11.notation)
````

## 3. Constructing $(|\Phi^+\rangle)$

The first Bell state is:

$
|\Phi^+\rangle =
\frac{|00\rangle + |11\rangle}{\sqrt{2}}
$

We can create it using two gates:

1. Apply Hadamard to the first qubit.
2. Apply CNOT with the first qubit as control and the second as target.

Circuit:

```text
q0 ───H───●───
          │
q1 ───────X───

````python
state = StateVector([
    Qubit.zero(),
    Qubit.zero()
])

state = state.apply(oqubit.H, [], [0])
state = state.apply(oqubit.CX, [0], [1])

print(state.notation)
print(state.state)
````

````python
phi_plus_expected = np.array([
    1 / np.sqrt(2),
    0,
    0,
    1 / np.sqrt(2)
], dtype=complex)

np.allclose(state.state, phi_plus_expected)
````

## 4. Constructing $(|\Phi^-\rangle)$

The second Bell state is:

$
|\Phi^-\rangle =
\frac{|00\rangle - |11\rangle}{\sqrt{2}}
$

We can create it by applying:

1. Hadamard to qubit 0
2. Pauli-Z to qubit 0
3. CNOT with qubit 0 as control and qubit 1 as target

Circuit:

```text
q0 ───H───Z───●───
              │
q1 ───────────X───
```

````python
state = StateVector([
    Qubit.zero(),
    Qubit.zero()
])

state = state.apply(oqubit.H, [], [0])
state = state.apply(oqubit.Z, [], [0])
state = state.apply(oqubit.CX, [0], [1])

print(state.notation)

print(state.state)
````

````python
phi_minus_expected = np.array([
    1 / np.sqrt(2),
    0,
    0,
    -1 / np.sqrt(2)
], dtype=complex)

np.allclose(state.state, phi_minus_expected)
````

## 5. Constructing $(|\Psi^+\rangle$)

The third Bell state is:

$
|\Psi^+\rangle =
\frac{|01\rangle + |10\rangle}{\sqrt{2}}
$

We start with:

$
|01\rangle
$

Then apply Hadamard to qubit 0:

$
|01\rangle
\rightarrow
\frac{|01\rangle+|11\rangle}{\sqrt2}
$

Finally, CNOT flips the second qubit whenever the first qubit is `1`:

$
\frac{|01\rangle+|11\rangle}{\sqrt2}
\rightarrow
\frac{|01\rangle+|10\rangle}{\sqrt2}
$

Circuit:

```text
q0 ───H───●───
          │
q1 ───────X───
```

````python
state = StateVector([
    Qubit.zero(),
    Qubit(alpha=0, beta=1)
])

state = state.apply(oqubit.H, [], [0])
state = state.apply(oqubit.CX, [0], [1])

print(state.notation)

print(state.state)
````

````python
psi_plus_expected = np.array([
    0,
    1 / np.sqrt(2),
    1 / np.sqrt(2),
    0
], dtype=complex)

np.allclose(state.state, psi_plus_expected)
````

## 6. Constructing $(|\Psi^-\rangle$)

The final Bell state is:

$
|\Psi^-\rangle =
\frac{|01\rangle - |10\rangle}{\sqrt2}
$

Starting from:

$
|01\rangle
$

we apply:

1. Hadamard to qubit 0
2. Pauli-Z to qubit 0
3. CNOT with qubit 0 as control and qubit 1 as target

Circuit:

```text
q0 ───H───Z───●───
              │
q1 ───────────X───
```

````python
state = StateVector([
    Qubit.zero(),
    Qubit(alpha=0, beta=1)
])

state = state.apply(oqubit.H, [], [0])
state = state.apply(oqubit.Z, [], [0])
state = state.apply(oqubit.CX, [0], [1])

print(state.notation)

print(state.state)
````

````python
psi_minus_expected = np.array([
    0,
    1 / np.sqrt(2),
    -1 / np.sqrt(2),
    0
], dtype=complex)

np.allclose(state.state, psi_minus_expected)
````

## 7. The Four Bell States

The Bell states can be summarized as:

| State | Mathematical form |
|---|---|
| $(|\Phi^+\rangle$) | $((|00\rangle + |11\rangle)/\sqrt2$) |
| $(|\Phi^-\rangle$) | $((|00\rangle - |11\rangle)/\sqrt2$) |
| $(|\Psi^+\rangle$) | $((|01\rangle + |10\rangle)/\sqrt2$) |
| $(|\Psi^-\rangle$) | $((|01\rangle - |10\rangle)/\sqrt2$) |

````python
bell_states = {
    "Phi+": phi_plus_expected,
    "Phi-": phi_minus_expected,
    "Psi+": psi_plus_expected,
    "Psi-": psi_minus_expected,
}

for name, vector in bell_states.items():
    print(f"{name}:")
    print(vector)
    print()
````

## 8. Bell-State Normalization

Every valid quantum state must satisfy:

$
\sum_i |\alpha_i|^2 = 1
$

For a Bell state:

$
\left|\frac{1}{\sqrt2}\right|^2
+
\left|\frac{1}{\sqrt2}\right|^2
=
1
$

Therefore every Bell state is normalized.

````python
for name, vector in bell_states.items():
    probability = np.sum(np.abs(vector) ** 2)
    print(f"{name}: {probability}")
````

## 9. Orthogonality

The four Bell states are mutually orthogonal.

For two different Bell states:

$
\langle B_i|B_j\rangle = 0
$

This means they form an orthonormal basis for the two-qubit Hilbert space.

````python
names = list(bell_states)
vectors = list(bell_states.values())

for i in range(len(vectors)):
    for j in range(i + 1, len(vectors)):
        inner_product = np.vdot(vectors[i], vectors[j])
        print(
            f"<{names[i]}|{names[j]}> = {inner_product}"
        )
````

## 10. Why Bell States Are Important

Bell states are entangled states.

For example:

$
|\Phi^+\rangle =
\frac{|00\rangle+|11\rangle}{\sqrt2}
$

The state is not simply a tensor product of two independent single-qubit states.

If the two qubits are measured in the computational basis, the outcomes are correlated:

- `00`
- `11`

The corresponding amplitudes have equal magnitude.

Bell states are therefore fundamental examples of quantum entanglement.

# Final Verification

## 12. Verify All Four Bell States

Let's verify that OQubit produces the expected vectors for all four Bell states.

````python
# |Phi+>
state = StateVector([Qubit.zero(), Qubit.zero()])
state = state.apply(oqubit.H, [], [0])
state = state.apply(oqubit.CX, [0], [1])

phi_plus_result = state.state.copy()


# |Phi->
state = StateVector([Qubit.zero(), Qubit.zero()])
state = state.apply(oqubit.H, [], [0])
state = state.apply(oqubit.Z, [], [0])
state = state.apply(oqubit.CX, [0], [1])

phi_minus_result = state.state.copy()


# |Psi+>
state = StateVector([
    Qubit.zero(),
    Qubit(alpha=0, beta=1)
])
state = state.apply(oqubit.H, [], [0])
state = state.apply(oqubit.CX, [0], [1])

psi_plus_result = state.state.copy()


# |Psi->
state = StateVector([
    Qubit.zero(),
    Qubit(alpha=0, beta=1)
])
state = state.apply(oqubit.H, [], [0])
state = state.apply(oqubit.Z, [], [0])
state = state.apply(oqubit.CX, [0], [1])

psi_minus_result = state.state.copy()
````

````python
print("Phi+ :", np.allclose(phi_plus_result, phi_plus_expected))
print("Phi- :", np.allclose(phi_minus_result, phi_minus_expected))
print("Psi+ :", np.allclose(psi_plus_result, psi_plus_expected))
print("Psi- :", np.allclose(psi_minus_result, psi_minus_expected))
````
