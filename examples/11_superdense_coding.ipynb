# Superdense Coding

Superdense coding allows two classical bits to be transmitted
using a shared entangled pair of qubits and a single transmitted qubit.

This example uses OQubit's built-in Superdense Coding algorithm.


```python
import oqubit
from oqubit.algorithms import encode,decode
```

## Encode and Decode

Encode the two classical bits and recover them using Superdense Coding.

```python
result = encode(1,0)
print("Input:   ", (1,0))
print("Encoded: ", result)
```

## Test All Messages

Superdense coding can represent all four possible two-bit messages.

```python
result =encode(0,0)
print((0,0), "->", result)
print(decode(result))
result =encode(0,1)
print((0,1), "->", result)
print(decode(result))
result =encode(1,0)
print((1,0), "->", result)
print(decode(result))
result =encode(1,1)
print((1,1), "->", result)
print(decode(result))
```
