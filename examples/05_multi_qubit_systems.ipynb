## 2. From One Qubit to Multiple Qubits

A single qubit has two computational basis states:

$$
|0\rangle,\quad |1\rangle
$$

A system containing two qubits has four:

$$
|00\rangle,\quad |01\rangle,\quad |10\rangle,\quad |11\rangle
$$

For three qubits there are eight:

$$
|000\rangle,\ |001\rangle,\ |010\rangle,\ |011\rangle,
|100\rangle,\ |101\rangle,\ |110\rangle,\ |111\rangle
$$

In general, an `n`-qubit system has:

$$
2^n
$$

computational-basis states.

## 5. Computational-Basis Ordering

For two qubits, OQubit uses the computational basis:

$$
|00\rangle,\ |01\rangle,\ |10\rangle,\ |11\rangle
$$

The corresponding state vector is:

$$
|\psi\rangle =
\begin{bmatrix}
c_{00}\\
c_{01}\\
c_{10}\\
c_{11}
\end{bmatrix}
$$

For example:

$$
|10\rangle =
\begin{bmatrix}
0\\
0\\
1\\
0
\end{bmatrix}
$$

```python
state = StateVector([
    Qubit.one(),
    Qubit.zero()
])

print(state.state)
```

````python
state = StateVector([
    Qubit.one(),
    Qubit.zero()
])

print(state.state)
````

## 6. Tensor Products

The multi-qubit state is constructed using the tensor product.

For two states:

$$
|\psi_1\rangle =
\begin{bmatrix}
a\\
b
\end{bmatrix}
$$

and

$$
|\psi_2\rangle =
\begin{bmatrix}
c\\
d
\end{bmatrix}
$$

their tensor product is:

$$
|\psi_1\rangle\otimes|\psi_2\rangle
=
\begin{bmatrix}
ac\\
ad\\
bc\\
bd
\end{bmatrix}
$$

Let's verify this numerically.

```python
q1 = Qubit(
    alpha=1 / np.sqrt(2),
    beta=1 / np.sqrt(2)
)

q2 = Qubit.zero()

state = StateVector([q1, q2])

print(state.state)
```

The result corresponds to:

$$
|+\rangle\otimes|0\rangle
=
\frac{1}{\sqrt2}
(|00\rangle+|10\rangle)
$$


````python
q1 = Qubit(
    alpha=1 / np.sqrt(2),
    beta=1 / np.sqrt(2)
)

q2 = Qubit.zero()

state = StateVector([q1, q2])

print(state.state)
````

## 8. Two-Qubit Superposition

Now create:

$$
|+\rangle\otimes|+\rangle
$$

```python
state = StateVector([
    Qubit.plus(),
    Qubit.plus()
])

print(state.state)
```

Mathematically:

$$
|+\rangle\otimes|+\rangle
=
\frac12
(
|00\rangle+
|01\rangle+
|10\rangle+
|11\rangle
)
$$

Calculate the probabilities:

```python
probabilities = np.abs(state.state) ** 2

print(probabilities)
print("Total probability:", probabilities.sum())
```

Each basis state has probability:

$$
\frac14
$$


````python
state = StateVector([
    Qubit.plus(),
    Qubit.plus()
])

print(state.state)

probabilities = np.abs(state.state) ** 2

print(probabilities)
print("Total probability:", probabilities.sum())
````

## 13. Arbitrary Product State

Each individual qubit can have a different state.

For example:

```python
q1 = Qubit(
    alpha=0.8,
    beta=0.6
)

q2 = Qubit.plus()

q3 = Qubit.minus()

state = StateVector([
    q1,
    q2,
    q3
])

print("State:")
print(state)

print("\nState vector:")
print(state.state)

print("\nNumber of amplitudes:")
print(len(state.state))
```

This creates the product state:

$$
|\psi\rangle
=
|\psi_1\rangle
\otimes
|+\rangle
\otimes
|-\rangle
$$


````python
q1 = Qubit(
    alpha=0.8,
    beta=0.6
)

q2 = Qubit.plus()

q3 = Qubit.minus()

state = StateVector([
    q1,
    q2,
    q3
])

print("State:")
print(state)

print("\nState vector:")
print(state.state)

print("\nNumber of amplitudes:")
print(len(state.state))
````

## 14. Checking Normalization

The norm of a valid state vector is:

$$
\lVert\psi\rVert =
\sqrt{\sum_i|c_i|^2}
$$

Calculate it:

```python
norm = np.linalg.norm(state.state)

print("State-vector norm:", norm)
```

For a normalized state, the result should be approximately:

```text
1.0
```

The total probability is:

```python
probability_sum = np.sum(np.abs(state.state) ** 2)

print("Total probability:", probability_sum)

````python
norm = np.linalg.norm(state.state)
print("State-vector norm:", norm)
probability_sum = np.sum(np.abs(state.state) ** 2)
print("Total probability:", probability_sum)
````

## 16. Important Distinction: Product vs Entangled States

The states constructed so far are **product states**.

For example:

$$
|\psi\rangle =
|\psi_1\rangle\otimes|\psi_2\rangle
$$

can be separated into independent single-qubit states.

Later, when we introduce controlled gates, we will construct states that cannot be represented as a simple tensor product of individual qubit states.

These are **entangled states**.

A famous example is the Bell state:

$$
|\Phi^+\rangle =
\frac{|00\rangle+|11\rangle}{\sqrt2}
$$

We will construct states like this in a later OQubit example.

## 18. Summary

A multi-qubit quantum system is represented by a single state vector.

For `n` qubits:

$$
|\psi\rangle =
\sum_{i=0}^{2^n-1}c_i|i\rangle
$$

and:

$$
\dim(|\psi\rangle)=2^n
$$

The key ideas are:

| Concept         | Meaning                                    | 
| --------------- | ------------------------------------------ |
| `Qubit`         | Single-qubit state                         |     
| `StateVector`   | Multi-qubit state                          |     
| Tensor product  | Combines quantum systems                   |   
| Basis state     | Computational state such as <div>"010"<div> |
| Amplitude       | Complex coefficient `cᵢ`                   |      
| Probability     | Calculate the probability of Qubit existing in different states|
| Product state   | Separates into individual qubit states     |       
| Entangled state | Cannot be separated into individual states |
